```python
#For numerical columns
df['Popularity'] = df['Popularity'].fillna(
    df['Popularity'].median()
)

df['Vote_Count'] = df['Vote_Count'].fillna(
    df['Vote_Count'].median()
)

df['Vote_Average'] = df['Vote_Average'].fillna(
    df['Vote_Average'].median()
)
```

```python
#For categorical columns
df['Original_Language'] = df['Original_Language'].fillna(
    df['Original_Language'].mode()[0]
)

df['Genre'] = df['Genre'].fillna(
    df['Genre'].mode()[0]
)
```

```python
#For text columns
df['Title'] = df['Title'].fillna('Unknown')

df['Overview'] = df['Overview'].fillna('No Overview')

df['Poster_Url'] = df['Poster_Url'].fillna('No URL')
```

```python
#Find outliers using IQR
#IQR = Q3 - Q1
#Lower Bound = Q1 - 1.5 × IQR
#Upper Bound = Q3 + 1.5 × IQR

column= ['popularity','Vote_Count','Vote_Average']
def find_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))

    return outliers
```

```python
top10_movies = df.nlargest(10, 'Popularity')

plt.figure(figsize=(10, 6))

plt.barh(top10_movies['Title'], top10_movies['Popularity'])

plt.title('Top 10 Most Popular Movies')
plt.xlabel('Popularity')
plt.ylabel('Movie')

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
genres = df['Genre'].dropna().str.split(',').explode()
genres = genres.str.strip()

top_genres = genres.value_counts().head(10)

plt.figure(figsize=(10, 6))

plt.bar(top_genres.index, top_genres.values)

plt.title('Top 10 Movie Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Movies')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
movies_per_year = df['Release_Year'].value_counts().sort_index()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(movies_per_year.index, movies_per_year.values)

plt.title('Movies Released Over Time')
plt.xlabel('Release Year')
plt.ylabel('Number of Movies')

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Vote_Average'], bins=20)

plt.title('Rating Distribution')
plt.xlabel('Vote Average')
plt.ylabel('Number of Movies')

plt.tight_layout()
plt.show()
```

```python
corr = df[
    ['Popularity', 'Vote_Count', 'Vote_Average']
].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.tight_layout()
plt.show()
```
